```python
# Tuples in python

'''
A tuple is a ordered collection of items, just like list.
- You can mutate a list
- Tuple is immutable (once you created a tuple you can not change, add, or delete item)
'''

fruits = ("apple", "orange", "banana");

print(fruits)
print(type(fruits))
```

Different Between List and Tuple

- List is Slower, Tuple is fast (optimized for memory)
- List is mutable, Tuples is Immutable
- List has many methods, Tuple only has count and index
- List is used for changing data, Tuple is for constant data

```python
# Accessing an item of tuple
print(fruits[0])
```

```python
# Trying to change an item
fruits[0] = "cheery"
```

```python
# tuple unpacking
a,o,b = fruits
print(a)
a = "mango"

print(a)
```

```python
# Tuple with only one item

'''
If you want to create tuple with only one item then you must include trailing comma
otherwise python confuses it with string and type will be string
'''

t1 = ("apple")
print(t1)
print(type(t1))

t2 = ("apple",)
print(t2)
print(type(t2))
```

# Dictionary

While lists and tuples are ordered sequences of single items,
a Dictionary in Python is a **collection of key-value pairs**.
Think of it like a real-world dictionary:
you look up a "word" (the key) to find its "definition" (the value).

```python
# Dictionaries in python
user = {
    "name": "Avinash",
    "age" : 25,
    "is_student": True,
    "skills" : ["Python", "ML"]
}

print(user)
print(type(user))
```

> **Rules for Keys**: Keys must be unique and immutable (strings, numbers, or tuples). Values can be absolutely anything-even lists or other dictionaries!

```python
# Accessing a value
print(user["name"])
```

```python
# Adding or updating item

# update age
user["age"] = 26

# add city

user["city"] = "Bengaluru"

print(user)
```

```python
# looping through dictionary

# Loop through keys

for key in user:
  print(key)
```

```python
# Loop through value
for value in user.values():
  print(value)
```

```python
# Loop through both key and value

for key, value in user.items():
  print(f"The {key} is {value}")
```

```python
# Get a key which is not present

print(user["test"])
```

```python
# Get a key safely if not exists then it will return None
print(user.get("test"))
```

```python
# Remove a key and return its value
city = user.pop("city")
print(city)
print(user)
```

```python
# check how many key value pair are inside
len(user)
```

# Sets

A Set in Python is an unordered collection of items where **every element must be unique**. If you try to add a duplicate item to a set, Python will simply ignore it. Think of it like a guest list where no one can be invited twice.

```python
# Creatinga set with duplicate elements
number_set = {1,2,3,1,2,3,3,4,4,5}

print(number_set)
print(type(number_set))
print(len(number_set))
```

> **The Empty Set**: If you want to create an empty set, you cannot use {} because Python will think it's an empty dictionary. You must use set() instead: empty_set = set().

```python
empty_set = {}

print(empty_set)
print(type(empty_set))

empty_set = set()
print(empty_set)
print(type(empty_set))
```

```python
# Modifying a set

fruits_set = {"apple", "orange"}

# Add a new fruit
fruits_set.add("banana")

# remove a fruit
fruits_set.remove("orange")

print(fruits_set)
```

```python
# Union of two set (to combine everything)

setA = {1,2,3,4,5}
setB = {4,5,6,7,8}

union_set = setA.union(setB)

print(union_set)
```

```python
# intersection (to find overlap)

intersection_set = setA.intersection(setB)

print(intersection_set)
```

```python
# difference (to find whats unique)
diff_set = setA.difference(setB) # it will find unique in set a
print(diff_set)
```

```python
# all unique item
diff_set_a = setA.difference(setB) # it will find unique in set b
diff_set_b = setB.difference(setA)

final_set = diff_set_a.union(diff_set_b)

print(final_set)
```

# Function

A function is a block of code that only runs when it is called. You define a function using the def keyword.

- **Parameters** are the placeholders you define in the function signature.

- **Arguments** are the actual values you pass into those placeholders.

- **Return** values use the return keyword to send data back to the caller. Default return value is None

```python
# Defining a function

def greet():
  print("hello")
```

```python
# Call function
greet()
```

```python
# function with parameters and return value
def greet_user(first_name, last_name):
  full_name = f"{first_name} {last_name}"
  return f"Hi {full_name}"
```

```python
# Call function
message = greet_user("Avinash", "Kumar")
print(message)
```

```python
# Default argument

def power(base, exponent=2):
  return base ** exponent;

print(power(2))
print(power(2,3))
```

```python
# Lambda function

def square(x):
  return x * x

square_lambda = lambda x: x * x

print(square(5))
print(square_lambda(5))
```

```python
# Lambda functions are used for quick transformations

scores = [("Math", 90), ("Science", 95), ("History", 70)]

scores.sort(key = lambda item: item[1])

print(scores)
```
